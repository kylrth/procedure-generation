```python
from collections import Counter
from pathlib import Path

from matplotlib import pyplot as plt

from all_compare import Comparison, OldSystemOption, PairwiseResult
from run import DatasetOption
```

```python
c = Comparison(ds=DatasetOption.LCStep, s1=OldSystemOption.RAGNoCritic, s2=OldSystemOption.AAG, logdir_root=Path("../output_linear_final"))
res1, res2, respw = c.collect()

pw_mode = Counter(respw).most_common(1)[0][0]
title_extra = ""
match pw_mode:
    case PairwiseResult.First:
        title_extra = f" (expect {c.s1} win)"
    case PairwiseResult.Second:
        title_extra = f" (expect {c.s2} win)"
    case _:
        print("neither won...")
```

```python
def bi_hist(eval_name: str):
        _, bins, _ = plt.hist(res1[eval_name], alpha=0.5, label=c.s1, bins=20)
        plt.hist(res2[eval_name], alpha=0.5, label=c.s2, bins=bins)
        plt.title(eval_name + title_extra)
        plt.legend()
```

```python
print(", ".join(res1.keys()))
```

```python
bi_hist("Overall")
plt.show()
```

```python
bi_hist("Api-Overlap")
plt.show()
```

```python
bi_hist("Inp-Used")
plt.show()
```
